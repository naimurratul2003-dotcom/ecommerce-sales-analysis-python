3. Which product category sells the most?

```python
df['product_category'].value_counts().plot.bar()
```

Electronics: 1,777 items (highest); Beauty: 723 items (lowest)
Electronics is top selling category with 35% orders. Beauty is lowest with 14%.

4. Sales by Region

```python
df.groupby('region')['revenue'].sum().plot.bar()
```

West: 1,289 orders, South: 1,208 orders—all regions are roughly equal.
Sales are balanced across all 4 regions. West leads slightly with 25.7% orders.

5. Impact of discounts – Does offering higher discounts increase revenue?

```python
sns.scatterplot(x='discount', y='revenue', data=df)
```

You will see that discounts range from 0% to 35%, but there is no direct relationship with revenue.

6. Delivery Days vs. Customer Rating – Do ratings drop with late deliveries?

```python
df.groupby('delivery_days')['customer_rating'].mean().plot()
```

I expected ratings to drop with delayed delivery, but the data showed otherwise.

- Average rating for 1-day delivery: 2.98
- Average rating for 11-day delivery: 2.89
- The difference is only 0.09!
- Correlation: -0.017, meaning there is no relationship.

Conclusion: For this company, there is no correlation between delivery time and customer rating.
Perhaps customers rate based on product quality rather than delivery time.
Or, customers accept delivery times ranging from 1 to 11 days.

7. Payment Method

```python

counts = df['payment_method'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(counts.values,
        labels=counts.index,
        autopct='%1.1f%%',
        startangle=90,
        colors=['#264653','#2a9d8f','#e9c46a'],
        wedgeprops={'edgecolor':'white', 'linewidth':2},
        textprops={'fontsize':12, 'weight':'bold'})

plt.title('Payment Method Distribution', weight='bold')
plt.show()
```

Card 45%, COD 35%, Wallet 19%
45% of customers prefer card payments, while COD stands at 35%.

8. Monthly Sales Trend (From Date)

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['month'] = df['order_date'].dt.month
df.groupby('month')['revenue'].sum().plot()
```

- Highest Revenue: May – $479,772 (479k)
- Lowest Revenue: April – $398,715 (398k)
- Difference: Sales in May were $81k (20%) higher than in April.

- Revenue remained stable between approximately $400k and $480k from January to December.
- There were no sudden, significant drops in any month, indicating consistent business performance throughout the year.

Conclusion: Sales peaked in May; there may have been a festival or a special offer during that month.
The company could apply the strategy used in May to other months as well.
